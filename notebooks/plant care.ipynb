## Step 24: Create a Flexible Plant Search

We improve the search system so that it can find plants even when the user enters only part of the plant name.

```python
def flexible_search(plant_name):
    
    plant_name = plant_name.lower().strip()
    
    result = master_df[
        master_df["typeName"]
        .str.lower()
        .str.contains(plant_name, na=False)
    ]
    
    return result[["typeName", "category"]]
```

## Step 32: Standardize User Plant Names

Users may enter common plant names that differ from the names stored in the Plantfolio dataset.

This function checks the plant alias dictionary and converts common names into the corresponding Plantfolio names.

```python
def standardize_plant_name(plant_name):
    
    # Convert user input to lowercase and remove extra spaces
    plant_name = plant_name.lower().strip()
    
    # Check if the name exists in the alias dictionary
    if plant_name in plant_aliases:
        return plant_aliases[plant_name]
    
    # If no alias is found, return the original name
    return plant_name
```

## Step 34: Create the Smart Plant Search System

The smart plant search system understands common plant names using the alias dictionary.

It first converts the user's input into a standardized plant name, then searches the Plantfolio dataset for an exact match.

If an exact match is not found, it performs a flexible search.

```python
def smart_search_plant(plant_name):
    
    # Standardize the plant name using aliases
    standardized_name = standardize_plant_name(plant_name)
    
    # Convert to lowercase
    search_name = standardized_name.lower()
    
    # Try to find an exact match
    exact_result = master_df[
        master_df["typeName"]
        .str.lower() == search_name
    ]
    
    # Return exact match if found
    if not exact_result.empty:
        return exact_result
    
    # If no exact match is found, use flexible search
    partial_result = master_df[
        master_df["typeName"]
        .str.lower()
        .str.contains(search_name, na=False)
    ]
    
    return partial_result
```

# Text Preprocessing

In this step, we clean the plant problem descriptions.

This helps make the text more consistent before converting it into numerical features.

```python
import re

def clean_text(text):

    # Convert text to lowercase
    text = text.lower()

    # Remove special characters and punctuation
    text = re.sub(r"[^a-zA-Z\s]", " ", text)

    # Remove extra spaces
    text = re.sub(r"\s+", " ", text).strip()

    return text
```

# Step 6: Improve Training Data

In this step, we improve the model's ability to understand different ways users may describe the same plant problem.

Only the training data is augmented.

The testing data remains unchanged so that the final evaluation remains fair.

```python
import pandas as pd
import random

# Set random seed for reproducible results

random.seed(42)


# ============================================================
# FUNCTION TO CREATE NATURAL LANGUAGE VARIATIONS
# ============================================================

def augment_description(text):

    variations = [text]

    # Possible natural prefixes users may use

    prefixes = [

        "My plant has this problem: ",

        "Can you help? ",

        "I noticed that ",

        "Please help, ",

        "I think something is wrong because ",

        "What could be wrong? ",

        "Recently, "

    ]


    # Choose one random prefix

    prefix = random.choice(prefixes)


    # Create augmented version

    augmented_text = prefix + text.lower()


    variations.append(augmented_text)


    return variations


# ============================================================
# AUGMENT TRAINING DATA ONLY
# ============================================================

augmented_X = []

augmented_y = []


for text, label in zip(X_train, y_train):

    # Create variations for each training example

    variations = augment_description(text)


    for variation in variations:

        augmented_X.append(variation)

        augmented_y.append(label)


# ============================================================
# CONVERT BACK TO PANDAS SERIES
# ============================================================

X_train_augmented = pd.Series(augmented_X)

y_train_augmented = pd.Series(augmented_y)


# ============================================================
# DISPLAY RESULTS
# ============================================================

print("Original training examples:", len(X_train))

print("Augmented training examples:", len(X_train_augmented))

print("\nTesting examples remain unchanged:", len(X_test))


# Display a few augmented examples

augmentation_preview = pd.DataFrame({

    "description": X_train_augmented[:10],

    "problem": y_train_augmented[:10]

})


augmentation_preview
```

# Step 7: TF-IDF Vectorization

In this step, we convert the plant problem descriptions from text into numerical features.

We fit the TF-IDF vectorizer only on the augmented training data.

The same fitted vectorizer is then used to transform the test data.

```python
from sklearn.feature_extraction.text import TfidfVectorizer


# ============================================================
# CREATE TF-IDF VECTORIZER
# ============================================================

tfidf_vectorizer = TfidfVectorizer(

    # Use individual words and two-word combinations
    ngram_range=(1, 2),

    # Keep words appearing at least once
    min_df=1,

    # Apply logarithmic scaling
    sublinear_tf=True,

    # Remove common English words
    stop_words="english"
)


# ============================================================
# FIT AND TRANSFORM AUGMENTED TRAINING DATA
# ============================================================

X_train_tfidf = tfidf_vectorizer.fit_transform(
    X_train_augmented
)


# ============================================================
# TRANSFORM TEST DATA
# ============================================================

X_test_tfidf = tfidf_vectorizer.transform(
    X_test
)


# ============================================================
# DISPLAY INFORMATION
# ============================================================

print(
    "Training feature matrix shape:",
    X_train_tfidf.shape
)

print(
    "\nTesting feature matrix shape:",
    X_test_tfidf.shape
)

print(
    "\nNumber of features:",
    len(tfidf_vectorizer.get_feature_names_out())
)
```

# Step 7: Train the Logistic Regression Model

In this step, we train a Logistic Regression model using the TF-IDF features.

The model will learn to classify plant problem descriptions into one of the 15 problem categories.

```python
from sklearn.linear_model import LogisticRegression


# Create the model

problem_model = LogisticRegression(
    max_iter=3000,
    class_weight="balanced",
    random_state=42
)


# Train the model
problem_model.fit(
    X_train_tfidf,
    y_train_augmented

)


print("Model training completed successfully!")
```

# Step 8: Evaluate the Model

In this step, we evaluate the trained Logistic Regression model using
the testing dataset.

# Step 11: Create a Plant Problem Prediction Function

This function accepts a user's description of a plant problem.

It cleans the text, converts it into TF-IDF features, predicts
the plant problem, and calculates the confidence score.

```python
# ============================================================
# STEP 11: CREATE PREDICTION FUNCTION
# ============================================================

def predict_plant_problem(user_description):


    # ========================================================
    # CLEAN USER INPUT
    # ========================================================

    cleaned_description = clean_text(
        user_description
    )


    # ========================================================
    # CONVERT TEXT TO TF-IDF FEATURES
    # ========================================================

    description_tfidf = tfidf_vectorizer.transform(
        [cleaned_description]
    )


    # ========================================================
    # PREDICT PROBLEM
    # ========================================================

    prediction = problem_model.predict(
        description_tfidf
    )[0]


    # ========================================================
    # GET CONFIDENCE SCORE
    # ========================================================

    probabilities = problem_model.predict_proba(
        description_tfidf
    )[0]


    confidence = (
        max(probabilities) * 100
    )


    # ========================================================
    # RETURN RESULTS
    # ========================================================

    return {

        "problem": prediction,

        "confidence": round(
            confidence,
            2
        )

    }
```

```python
# ============================================================
# TEST THE PREDICTION FUNCTION
# ============================================================

test_description = (
    "My plant has yellow leaves and the soil is always wet"
)


result = predict_plant_problem(
    test_description
)


print("Predicted Problem:")

print(
    result["problem"]
)


print("\nConfidence:")

print(
    result["confidence"],
    "%"
)
```

# Step 13: Complete Plant Problem Prediction Function

This function accepts a user's plant problem description and returns:

- Predicted plant problem
- Confidence score
- Possible cause
- Recommendation

```python
# ============================================================
# STEP 13: COMPLETE PREDICTION FUNCTION
# ============================================================

def predict_plant_problem(user_description):


    # ========================================================
    # CLEAN USER INPUT
    # ========================================================

    cleaned_description = clean_text(
        user_description
    )


    # ========================================================
    # CONVERT TEXT TO TF-IDF FEATURES
    # ========================================================

    description_tfidf = tfidf_vectorizer.transform(
        [cleaned_description]
    )


    # ========================================================
    # PREDICT PLANT PROBLEM
    # ========================================================

    prediction = problem_model.predict(
        description_tfidf
    )[0]


    # ========================================================
    # GET CONFIDENCE SCORE
    # ========================================================

    probabilities = problem_model.predict_proba(
        description_tfidf
    )[0]


    confidence = (
        max(probabilities) * 100
    )


    # ========================================================
    # GET PROBLEM INFORMATION
    # ========================================================

    problem_details = problem_information.get(
        prediction,
        {}
    )


    # ========================================================
    # GET CAUSE
    # ========================================================

    cause = problem_details.get(
        "cause",
        "Cause information is not available."
    )


    # ========================================================
    # GET RECOMMENDATION
    # ========================================================

    recommendation = problem_details.get(
        "recommendation",
        "Recommendation information is not available."
    )


    # ========================================================
    # RETURN COMPLETE RESULT
    # ========================================================

    return {

        "problem": prediction,

        "confidence": round(
            confidence,
            2
        ),

        "cause": cause,

        "recommendation": recommendation
    }
```
